```python
def func_calc(my_dict):
  total = 0
  count = 0
  for val in my_dict.values():
    try:
      total += val
      count += 1
    except TypeError:
      print(f"Ошибка! Значение '{val}' не является числом")
  try:
    average = total / count
  except ZeroDivisionError:
    print("Словарь должен быть не пустой!")
    average = 0
  return total,average
dct_3 = {}
print(func_calc(dct_3))
```

```python
def func_calc(my_dict,thr=0):
  total = 0
  count = 0
  for val in my_dict.values():
    try:
      if val > thr:
        total += val
        count += 1
    except TypeError:
      print(f"Ошибка! Значение '{val}' не является числом")
  try:
    average = total / count
  except ZeroDivisionError:
    print("Словарь должен быть не пустой!")
    average = 0
  return total,average
dct_1 = {'a':10, 'b':20, 'c':30}
print(func_calc(dct_1, 15))
```

```python
def filter_list(my_list, left=2, right=5):
  result=[]
  for val in my_list:
    if val >= left and val <= right:
      result.append(val)
  return result
test_list = [2,3,4,5,6,7,8,9]
print(filter_list(test_list))
```
